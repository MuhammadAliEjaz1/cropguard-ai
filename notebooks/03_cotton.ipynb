```python
CLASS_NAME_MAP = {
    "healthy":           "Healthy",
    "bacterialblight":   "BacterialBlight",
    "bacterial_blight":  "BacterialBlight",
    "bacterial blight":  "BacterialBlight",
    "curlvirus":         "CurlVirus",
    "curl_virus":        "CurlVirus",
    "curl virus":        "CurlVirus",
    "fusariumwilt":      "FusariumWilt",
    "fusarium_wilt":     "FusariumWilt",
    "fusarium wilt":     "FusariumWilt",
    "fussariumwilt":     "FusariumWilt",
    "fussarium_wilt":    "FusariumWilt",
}
 
def normalize_class_name(folder_name):
    key = folder_name.strip().lower().replace("-", "").replace(" ", "").replace("_", "")
    for map_key, final_name in CLASS_NAME_MAP.items():
        if map_key.replace("_", "").replace(" ", "") == key:
            return final_name
    print(f"  [WARNING] Unrecognized class folder name: '{folder_name}' -- kept as-is")
    return folder_name.strip()
```

```python
def find_class_folders(root_dir, skip_names=("train", "valid", "test", "images", "cotton", "cotton-original-augmented")):
    """
    Walks the directory tree and maps every folder containing images to a
    normalized class name. Handles three structural patterns seen across
    our cotton sources:
      1. Flat:                ClassName/*.jpg
      2. Split:                train/ClassName/*.jpg
      3. Severity (Mendeley):  ClassName/severity_level/*.jpg            <- RAW, keep
                               ClassName/severity_level/augment_result/  <- AUGMENTED, SKIP
    """
    severity_words = ("mild", "moderate", "severe", "critical", "early", "late")
    class_folders = {}
 
    for dirpath, dirnames, filenames in os.walk(root_dir):
        image_files = [f for f in filenames if f.lower().endswith((".jpg", ".jpeg", ".png"))]
        if len(image_files) == 0:
            continue
 
        folder_name = os.path.basename(dirpath)
 
        # SKIP any folder literally named "augment_result" -- these are
        # synthetic augmented duplicates, not genuinely new raw images.
        if folder_name.lower() == "augment_result":
            continue
 
        parent_name  = os.path.basename(os.path.dirname(dirpath))
        gparent_name = os.path.basename(os.path.dirname(os.path.dirname(dirpath)))
 
        if folder_name.lower() in severity_words and parent_name:
            effective_name = parent_name
        elif parent_name.lower() in severity_words and gparent_name:
            effective_name = gparent_name
        elif folder_name.lower() in skip_names:
            continue
        else:
            effective_name = folder_name
 
        normalized = normalize_class_name(effective_name)
        class_folders.setdefault(normalized, []).append(dirpath)
 
    return class_folders
 
print("=== Scanning Roboflow Source A (cotton-disease-detectionv2) ===")
roboflow_a_classes = find_class_folders(dataset_a.location)
for cls, folders in roboflow_a_classes.items():
    total = sum(len([f for f in os.listdir(p) if f.lower().endswith((".jpg",".jpeg",".png"))]) for p in folders)
    print(f"  {cls}: {total} images across {len(folders)} folder(s)")
 
print("\n=== Scanning Roboflow Source B (cotton-leaf-disease-gwaqa) ===")
roboflow_b_classes = find_class_folders(dataset_b.location)
for cls, folders in roboflow_b_classes.items():
    total = sum(len([f for f in os.listdir(p) if f.lower().endswith((".jpg",".jpeg",".png"))]) for p in folders)
    print(f"  {cls}: {total} images across {len(folders)} folder(s)")
 
print("\n=== Scanning existing Kaggle (seroshkarim) dataset ===")
existing_classes = find_class_folders(EXISTING_COTTON_DIR)
for cls, folders in existing_classes.items():
    total = sum(len([f for f in os.listdir(p) if f.lower().endswith((".jpg",".jpeg",".png"))]) for p in folders)
    print(f"  {cls}: {total} images across {len(folders)} folder(s)")
 
print("\n=== Scanning Mendeley SAR-CLD-2024 dataset ===")
mendeley_classes = find_class_folders(MENDELEY_COTTON_DIR)
for cls, folders in mendeley_classes.items():
    total = sum(len([f for f in os.listdir(p) if f.lower().endswith((".jpg",".jpeg",".png"))]) for p in folders)
    print(f"  {cls}: {total} images across {len(folders)} folder(s)")
```

```python
import hashlib
 
def file_hash(filepath):
    h = hashlib.md5()
    with open(filepath, "rb") as f:
        h.update(f.read(65536))
    return h.hexdigest()
 
def hash_all_images(folder_list):
    hashes = {}
    for folder in folder_list:
        for fname in os.listdir(folder):
            if fname.lower().endswith((".jpg", ".jpeg", ".png")):
                fpath = os.path.join(folder, fname)
                try:
                    hashes[file_hash(fpath)] = fpath
                except Exception:
                    pass
    return hashes
 
all_classes = set(
    list(roboflow_a_classes.keys()) +
    list(roboflow_b_classes.keys()) +
    list(existing_classes.keys()) +
    list(mendeley_classes.keys())
)
 
print("=== Duplicate check across all 4 sources ===")
for cls in sorted(all_classes):
    existing_h   = hash_all_images(existing_classes.get(cls, []))
    roboflow_a_h = hash_all_images(roboflow_a_classes.get(cls, []))
    roboflow_b_h = hash_all_images(roboflow_b_classes.get(cls, []))
    mendeley_h   = hash_all_images(mendeley_classes.get(cls, []))
 
    print(f"\n{cls}:")
    print(f"  Existing: {len(existing_h)}  |  Roboflow-A: {len(roboflow_a_h)}  |  "
          f"Roboflow-B: {len(roboflow_b_h)}  |  Mendeley: {len(mendeley_h)}")
 
    a_dupe_existing  = set(roboflow_a_h.keys()) & set(existing_h.keys())
    b_dupe_existing  = set(roboflow_b_h.keys()) & set(existing_h.keys())
    b_dupe_a         = set(roboflow_b_h.keys()) & set(roboflow_a_h.keys())
    md_dupe_existing = set(mendeley_h.keys()) & set(existing_h.keys())
    md_dupe_a        = set(mendeley_h.keys()) & set(roboflow_a_h.keys())
    md_dupe_b        = set(mendeley_h.keys()) & set(roboflow_b_h.keys())
 
    print(f"  Roboflow-A dupes vs Existing: {len(a_dupe_existing)}")
    print(f"  Roboflow-B dupes vs Existing: {len(b_dupe_existing)}")
    print(f"  Roboflow-B dupes vs Roboflow-A: {len(b_dupe_a)}")
    print(f"  Mendeley dupes vs Existing: {len(md_dupe_existing)}")
    print(f"  Mendeley dupes vs Roboflow-A: {len(md_dupe_a)}")
    print(f"  Mendeley dupes vs Roboflow-B: {len(md_dupe_b)}")
```

```python
import shutil
 
FINAL_DIR = "/kaggle/working/cotton_merged"
os.makedirs(FINAL_DIR, exist_ok=True)
 
final_counts = {}
 
for cls in sorted(all_classes):
    cls_out_dir = os.path.join(FINAL_DIR, cls)
    os.makedirs(cls_out_dir, exist_ok=True)
 
    seen_hashes = set()
    counts = {"existing": 0, "roboflow_a": 0, "roboflow_b": 0, "mendeley": 0}
 
    sources_in_order = [
        ("existing",   existing_classes.get(cls, [])),
        ("roboflow_a", roboflow_a_classes.get(cls, [])),
        ("roboflow_b", roboflow_b_classes.get(cls, [])),
        ("mendeley",   mendeley_classes.get(cls, [])),
    ]
 
    for source_name, folders in sources_in_order:
        for folder in folders:
            for fname in os.listdir(folder):
                if fname.lower().endswith((".jpg", ".jpeg", ".png")):
                    src = os.path.join(folder, fname)
                    try:
                        h = file_hash(src)
                    except Exception:
                        continue
                    if h not in seen_hashes:
                        dst = os.path.join(cls_out_dir, f"{source_name}_{counts[source_name]}_{fname}")
                        shutil.copy2(src, dst)
                        seen_hashes.add(h)
                        counts[source_name] += 1
 
    total = sum(counts.values())
    final_counts[cls] = total
    print(f"{cls}: existing={counts['existing']}, roboflow_a={counts['roboflow_a']}, "
          f"roboflow_b={counts['roboflow_b']}, mendeley={counts['mendeley']}  => TOTAL={total}")
 
print("\n=== FINAL MERGED COTTON DATASET ===")
total_all = sum(final_counts.values())
for cls, count in sorted(final_counts.items()):
    print(f"  {cls}: {count} images")
print(f"\nTOTAL: {total_all} images across {len(final_counts)} classes")
print(f"Saved to: {FINAL_DIR}")
```

```python
import tensorflow as tf
 
def validate_images(root_dir):
    removed = 0
    checked = 0
    for cls in os.listdir(root_dir):
        cls_dir = os.path.join(root_dir, cls)
        if not os.path.isdir(cls_dir):
            continue
        for fname in os.listdir(cls_dir):
            fpath = os.path.join(cls_dir, fname)
            checked += 1
            try:
                img_bytes = tf.io.read_file(fpath)
                tf.image.decode_image(img_bytes)
            except Exception:
                os.remove(fpath)
                removed += 1
    print(f"Checked {checked} images, removed {removed} corrupt files.")
 
print("\n=== Validating merged cotton dataset ===")
validate_images(FINAL_DIR)
 
print("\n=== FINAL counts after validation ===")
for cls in sorted(os.listdir(FINAL_DIR)):
    cls_dir = os.path.join(FINAL_DIR, cls)
    n = len([f for f in os.listdir(cls_dir) if f.lower().endswith((".jpg",".jpeg",".png"))])
    print(f"  {cls}: {n} images")
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications.efficientnet import EfficientNetB0, preprocess_input
import numpy as np
import json
import os
 
print("TensorFlow version:", tf.__version__)
print("GPU available:", tf.config.list_physical_devices('GPU'))
 
DATA_DIR = "/kaggle/working/cotton_merged"
IMG_SIZE = (224, 224)
BATCH_SIZE = 128
SEED = 42
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    validation_split=0.2,
    subset="training",
    seed=SEED,
)
 
val_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    validation_split=0.2,
    subset="validation",
    seed=SEED,
)
 
class_names = train_ds.class_names
NUM_CLASSES = len(class_names)
print("\nClasses found:", class_names)
print("Number of classes:", NUM_CLASSES)
 
# Save class names immediately so we don't lose this mapping
with open("/kaggle/working/cotton_class_names.json", "w") as f:
    json.dump(class_names, f, indent=2)
```

```python
from sklearn.utils.class_weight import compute_class_weight
 
# Re-walk the directory to get raw label counts for class_weight computation
labels_flat = []
for cls_idx, cls_name in enumerate(class_names):
    cls_dir = os.path.join(DATA_DIR, cls_name)
    n_files = len([f for f in os.listdir(cls_dir) if f.lower().endswith((".jpg", ".jpeg", ".png"))])
    labels_flat.extend([cls_idx] * n_files)
 
labels_flat = np.array(labels_flat)
class_weight_values = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(labels_flat),
    y=labels_flat
)
class_weight_dict = {i: w for i, w in enumerate(class_weight_values)}
 
print("Class weights:")
for idx, cls_name in enumerate(class_names):
    print(f"  {cls_name}: {class_weight_dict[idx]:.3f}")
```

```python
augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal_and_vertical"),
    layers.RandomRotation(0.3),
    layers.RandomZoom(0.2),
    layers.RandomBrightness(0.2),
    layers.RandomContrast(0.2),
], name="augmentation")
 
def prepare_train(image, label):
    image = augmentation(image)
    image = preprocess_input(image)
    return image, label
 
def prepare_val(image, label):
    image = preprocess_input(image)
    return image, label
 
train_ds_prepared = train_ds.map(prepare_train, num_parallel_calls=tf.data.AUTOTUNE)
val_ds_prepared   = val_ds.map(prepare_val, num_parallel_calls=tf.data.AUTOTUNE)
 
train_ds_prepared = train_ds_prepared.prefetch(tf.data.AUTOTUNE)
val_ds_prepared   = val_ds_prepared.prefetch(tf.data.AUTOTUNE)
```

```python
base_model = EfficientNetB0(
    include_top=False,
    weights="imagenet",
    input_shape=(224, 224, 3)
)
base_model.trainable = False  # Phase 1: frozen base
 
inputs = keras.Input(shape=(224, 224, 3))
x = base_model(inputs, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.BatchNormalization()(x)
x = layers.Dropout(0.4)(x)
outputs = layers.Dense(NUM_CLASSES, activation="softmax")(x)
 
model = keras.Model(inputs, outputs, name="cotton_efficientnetb0")
model.summary()
```

```python
early_stop = keras.callbacks.EarlyStopping(
    monitor="val_accuracy",
    patience=3,
    restore_best_weights=True
)
 
checkpoint = keras.callbacks.ModelCheckpoint(
    "/kaggle/working/cotton_best.keras",
    monitor="val_accuracy",
    save_best_only=True
)
 
reduce_lr = keras.callbacks.ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.3,
    patience=2,
    min_lr=1e-6)
```

```python
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-3),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
 
print("=== PHASE 1: Training classification head (base frozen) ===")
history1 = model.fit(
    train_ds_prepared,
    validation_data=val_ds_prepared,
    epochs=5,
    class_weight=class_weight_dict,
    callbacks=[early_stop, checkpoint, reduce_lr]
)
```

```python
base_model.trainable = True
 
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-4),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
 
print("=== PHASE 2: Full fine-tuning (lr=1e-4) ===")
history2 = model.fit(
    train_ds_prepared,
    validation_data=val_ds_prepared,
    epochs=10,
    class_weight=class_weight_dict,
    callbacks=[early_stop, checkpoint, reduce_lr]
)
```

```python
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=3e-5),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
 
print("=== PHASE 3: Extended fine-tuning (lr=3e-5) ===")
history3 = model.fit(
    train_ds_prepared,
    validation_data=val_ds_prepared,
    epochs=15,
    class_weight=class_weight_dict,
    callbacks=[early_stop, checkpoint, reduce_lr]
)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
 
y_true = []
y_pred = []
 
for images, labels in val_ds_prepared:
    preds = model.predict(images, verbose=0)
    y_true.extend(np.argmax(labels.numpy(), axis=1))
    y_pred.extend(np.argmax(preds, axis=1))
 
y_true = np.array(y_true)
y_pred = np.array(y_pred)
 
print("=== Classification Report ===")
print(classification_report(y_true, y_pred, target_names=class_names, digits=4))
 
# Confusion matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap="Greens", xticklabels=class_names, yticklabels=class_names)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Cotton Disease Classification — Confusion Matrix")
plt.tight_layout()
plt.savefig("/kaggle/working/cotton_confusion_matrix.png", dpi=150)
plt.show()
 
print("\nFinal validation accuracy:", round(max(history3.history['val_accuracy']) * 100, 2), "%")
```
